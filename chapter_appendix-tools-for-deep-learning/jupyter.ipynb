Import the Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

```

```python
#importing the data
credit_card_data = pd.read_csv('/content/creditcard.csv')
```

```python
credit_card_data.head()
```

```python
credit_card_data.tail()
```

```python
credit_card_data.info()
```

```python
credit_card_data.isnull().sum()
```

```python
credit_card_data = credit_card_data.dropna()
```

```python
credit_card_data.isnull().sum()
```

```python
#Distribution of legit and fraudulent transactions
credit_card_data['Class'].value_counts()
```

This Data is unbalanced......... 0--> Normal Transaction 1--> Fraudulent Transaction

```python
#separating the data for analysis
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]
```

```python
print(legit.shape)
print(fraud.shape)
```

```python
#statistical measures
legit.Amount.describe()
```

```python
fraud.Amount.describe()
```

```python
credit_card_data.groupby('Class').mean()
```

Under-sampling

Build a sample dataset containing similar distribution of normal and fraud transactions

no. of fraudulent transaction--> 492

```python
legit_sample = legit.sample(n=492)
```

```python
new_dataset = pd.concat([legit_sample, fraud], axis=0)
```

```python
new_dataset.head()
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

Splitting the data into Features and Targets

```python
x= new_dataset.drop(columns='Class', axis=1)
y= new_dataset['Class']
```

```python
print(x)
```

```python
print(y)
```

Split the data into training and testing data

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y, random_state=2)
```

### Data Standardization

Standardizing the data is crucial for many machine learning algorithms, including Logistic Regression, as it ensures that all features contribute equally to the distance calculation. This helps the optimization algorithm converge more efficiently.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
print(x_train)
```

```python
print(x.shape, x_train.shape, x_test.shape)
```

Model Training

Logistic Regression

```python
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(x_train, y_train)
```

Model Evaluation----------Accuracy Score

```python
#Accuracy on training data
x_train_prediction = model.predict(x_train)
training_data_accuracy = accuracy_score(x_train_prediction, y_train)
```

```python
print('Accuracy on Training data : ', training_data_accuracy)
```

```python
#accuracy on test data
x_test_prediction = model.predict(x_test)
test_data_accuracy = accuracy_score(x_test_prediction, y_test)
```

```python
print('Accuracy on Training data : ', test_data_accuracy)
```
